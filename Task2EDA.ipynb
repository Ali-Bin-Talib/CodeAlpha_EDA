```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
print("Apni CSV file upload karein:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("\n--- TASK 2: Exploratory Data Analysis (EDA) ---")
print("\n[1] Dataset ki Pehli 5 Rows:")
display(df.head())
print("\n[2] Dataset ki Information & Data Types:")
print(df.info())
print("\n[3] Missing Values Check:")
print(df.isnull().sum())
print("\n[4] Basic Summary:")
print(f"Total Quotes / Rows: {len(df)}")
print(f"Unique Authors ki tadad: {df['Author'].nunique()}")
print("\n[5] Top 5 Authors with Most Quotes:")
top_authors = df['Author'].value_counts().head(5)
print(top_authors)
plt.figure(figsize=(10, 5))
top_authors.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Top 5 Authors by Number of Quotes')
plt.xlabel('Authors')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
summary_filename = "CodeAlpha_EDA_Summary.csv"
df.to_csv(summary_filename, index=False)
print(f"\nEDA is now ready! Summary file '{summary_filename}' file is save now .")
```
